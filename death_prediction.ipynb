```python
# linear regression 
from sklearn.model_selection import train_test_split
```

```python
maha
```

```python
# converting date and time to ordinal
maha['Date']=maha['Date'].map(dt.datetime.toordinal)
maha.head() 
```

```python
#getting  dependent variable  and independent variable
x = maha['Date']
y= maha['Confirmed']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
x_train
```

```python
y_train
```

```python
lr.fit(np.array(x_train).reshape(-1,1),np.array(y_train).reshape(-1,1))
```

```python
maha.tail()
```

```python
y_pred = lr.predict(np.array(x_test).reshape(-1,1))
```

```python
lr.predict(np.array([[737620]]))
```

```python
lr.predict(np.array([[737630]]))
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Create predictions
y_pred = lr.predict(np.array(x_test).reshape(-1,1))

# Calculate REGRESSION metrics
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Model Evaluation Metrics:")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"R² Score: {r2:.4f}")

# R² Score interpretation
if r2 > 0.8:
    print("Model performance: Excellent!")
elif r2 > 0.6:
    print("Model performance: Good")
elif r2 > 0.4:
    print("Model performance: Average")
else:
    print("Model performance: Needs improvement")
```

```python
# Simple accuracy percentage
accuracy_percentage = r2 * 100
print(f"\nModel Accuracy: {accuracy_percentage:.2f}%")
```

```python
# Actual vs Predicted visualization
plt.figure(figsize=(15,8))
plt.scatter(x_test, y_test, color='blue', label='Actual Values', alpha=0.6, s=100)
plt.scatter(x_test, y_pred, color='red', label='Predicted Values', alpha=0.6, s=100)
plt.plot(x_test, y_pred, color='red', alpha=0.3, linewidth=2)
plt.xlabel('Date (Ordinal)', fontsize=12)
plt.ylabel('Confirmed Cases', fontsize=12)
plt.title(f'Actual vs Predicted - Model Accuracy: {r2*100:.2f}%', fontsize=14)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Perfect prediction line
plt.figure(figsize=(10,10))
plt.scatter(y_test, y_pred, alpha=0.6)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel('Actual Values', fontsize=12)
plt.ylabel('Predicted Values', fontsize=12)
plt.title(f'Perfect Prediction Line - R² = {r2:.4f}', fontsize=14)
plt.grid(True, alpha=0.3)
plt.show()
```
